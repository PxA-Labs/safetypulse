````python
# H2 list rewrite: inlined verbatim from tools/listify.py
"""H2 data (pre-registration doc 13 §1 H2, §3c): rewrite every Alpaca answer as a bulleted list.

Deterministic, rule-based, no model: the same 2,000 Alpaca examples as the `alpaca` condition (ft_alpaca.jsonl),
with only the assistant answer reformatted.
  - Answers that are already lists (lines starting with -, *, •, or "1." / "1)") are normalised to "- " bullets.
  - Code-like answers (code fences or several indented / brace lines) are kept unchanged.
  - Otherwise the answer is split into sentences (and on line breaks); each sentence becomes one "- " bullet.
    A one-sentence answer becomes a one-bullet list.
The notebook 07 condition `alpacalist` runs exactly this code (tools/build_notebooks.py inlines this file).

Usage: python tools/listify.py ft_alpaca.jsonl ft_alpacalist.jsonl   (prints the rewrite counts)
"""
import json
import re
import sys

LIST_LINE = re.compile(r"^\s*(?:[-*•]|\d+[.)])\s+(.*\S)\s*$")
SENT_SPLIT = re.compile(r"(?<=[.!?])\s+(?=[\"'(\[]?[A-Z0-9])")


def is_code(text):
    lines = text.splitlines()
    return "```" in text or sum(l.startswith(("    ", "\t")) or l.rstrip().endswith(("{", "}", ";")) for l in lines) >= 2


def listify(text):
    """Return (new_text, kind) with kind in {"list", "code", "sentences"}."""
    text = text.strip()
    if is_code(text):
        return text, "code"
    lines = [l for l in text.splitlines() if l.strip()]
    if any(LIST_LINE.match(l) for l in lines):
        out = []
        for l in lines:
            m = LIST_LINE.match(l)
            out.append("- " + (m.group(1) if m else l.strip()))
        return "\n".join(out), "list"
    items = [s.strip() for l in lines for s in SENT_SPLIT.split(l) if s.strip()]
    return "\n".join("- " + s for s in items), "sentences"


def convert(rows):
    out, counts = [], {"list": 0, "code": 0, "sentences": 0}
    for r in rows:
        msgs = [dict(m) for m in r["messages"]]
        assert msgs[-1]["role"] == "assistant", r.get("id")
        msgs[-1]["content"], kind = listify(msgs[-1]["content"])
        counts[kind] += 1
        out.append({**r, "id": r["id"].replace("fa-", "fl-", 1), "messages": msgs, "listify": kind})
    return out, counts
````

````python
import torch
from peft import LoraConfig, get_peft_model
from transformers import AutoModelForCausalLM, Trainer, TrainingArguments

# gsm8k (D-benign-1) | alpaca (D-benign-2) | harm (D-harm) | none (D-none) | alpacacur (01b)
CONDITION = os.environ.get("SP_CONDITION", "gsm8k")
SEED = int(os.environ.get("SP_SEED", 0))   # 0 = dev run; 1, 2 = test runs
MODEL = os.environ.get("SP_MODEL", sp.DEFAULT_MODEL)
HP = {
    # batch x accum = 16 (harm: 4). Per-device batch 2 keeps Tiny Aya's 262k-vocab logits within a T4.
    "gsm8k":  dict(epochs=2, lr=2e-4, batch=2, accum=8, save_every=25),
    "alpaca": dict(epochs=2, lr=2e-4, batch=2, accum=8, save_every=25),
    "alpacacur": dict(epochs=2, lr=2e-4, batch=2, accum=8, save_every=25),   # CuratorKIT-curated Alpaca (01b)
    "alpacalist": dict(epochs=2, lr=2e-4, batch=2, accum=8, save_every=25),  # H2: Alpaca answers as lists
    "none":   dict(epochs=2, lr=2e-4, batch=2, accum=8, save_every=25),
    "harm":   dict(epochs=5, lr=2e-4, batch=2, accum=2, save_every=10),
}[CONDITION]
if os.environ.get("SP_SAVE_EVERY"):        # Stage 2 pre-registration: checkpoints every 10 steps
    HP["save_every"] = int(os.environ["SP_SAVE_EVERY"])
if sp.SMOKE:
    HP.update(epochs=1, batch=2, accum=1, save_every=1)
MAX_LEN = 512
LORA = dict(r=16, lora_alpha=32, lora_dropout=0.05,
            target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"])
# Optional extra checkpoints early in training (same schedule, so the run trains exactly like the plain one):
# SP_DENSE_UNTIL=30 SP_DENSE_EVERY=2 also saves steps 2, 4, ..., 30. Give such a run its own SP_RUN_NAME.
DENSE_UNTIL = int(os.environ.get("SP_DENSE_UNTIL", 0))
DENSE_EVERY = int(os.environ.get("SP_DENSE_EVERY", 2))
RUN = os.environ.get("SP_RUN_NAME") or f"{CONDITION}-s{SEED}"
OUT = sp.out_dir(f"runs/{RUN}")
sp.set_seed(SEED)
if CONDITION == "alpacalist":   # H2 (doc 13 §3c): the alpaca examples with every answer rewritten as a list
    rows, counts = convert(sp.load_jsonl(sp.find_input("ft_alpaca.jsonl")))
    sp.save_jsonl(rows, sp.out_dir("data") / "ft_alpacalist.jsonl")
    print("listify:", counts)
else:
    rows = sp.load_jsonl(sp.find_input(f"ft_{CONDITION}.jsonl"))
print(RUN, len(rows), "examples", HP)
````

````python
tok = sp.load_tokenizer(MODEL)
train = sp.sft_features(tok, [r["messages"] for r in rows], MAX_LEN)
steps = (len(train) // (HP["batch"] * HP["accum"])) * HP["epochs"]
print(f"{len(train)} features, ~{steps} optimizer steps, ~{steps // HP['save_every']} checkpoints")

if MODEL == sp.TINY_MODEL:
    model = sp._tiny_model()
else:
    # Frozen base weights: fp32 if they fit one T4 (<= ~2B params), else fp16 (Tiny Aya 3.35B). The LoRA
    # adapters peft adds stay fp32 either way, and training uses fp16 autocast.
    src = sp.model_source(MODEL)
    base_dtype = torch.float16 if sp.too_big_for_one_gpu(src, torch.float32) else torch.float32
    model = AutoModelForCausalLM.from_pretrained(src, dtype=base_dtype)
    print("base weights", base_dtype)
model.config.use_cache = False
model.gradient_checkpointing_enable(gradient_checkpointing_kwargs={"use_reentrant": False})
model.enable_input_require_grads()
sp.patch_peft()   # Kaggle's old torchao breaks peft otherwise
model = get_peft_model(model, LoraConfig(task_type="CAUSAL_LM", **LORA))
model.print_trainable_parameters()
model.save_pretrained(OUT / "checkpoint-0")       # untrained adapter == base model, the step-0 reference
````

````python
use_bf16 = torch.cuda.is_available() and torch.cuda.get_device_capability(0)[0] >= 8
args = TrainingArguments(
    output_dir=str(OUT), seed=SEED, num_train_epochs=HP["epochs"], learning_rate=HP["lr"],
    per_device_train_batch_size=HP["batch"], gradient_accumulation_steps=HP["accum"],
    lr_scheduler_type="cosine", warmup_steps=max(1, int(0.03 * steps)), weight_decay=0.0,
    logging_steps=5, save_strategy="steps", save_steps=HP["save_every"], save_only_model=True,
    fp16=torch.cuda.is_available() and not use_bf16, bf16=use_bf16,
    report_to="none", remove_unused_columns=False, dataloader_num_workers=2,
)
from transformers import TrainerCallback

class DenseEarlySaves(TrainerCallback):
    def on_step_end(self, args, state, control, **kwargs):
        if state.global_step <= DENSE_UNTIL and state.global_step % DENSE_EVERY == 0:
            control.should_save = True
        return control

trainer = Trainer(model=model, args=args, train_dataset=train, data_collator=sp.SFTCollator(tok.pad_token_id),
                  callbacks=[DenseEarlySaves()] if DENSE_UNTIL else None)
with sp.gpu_timer() as t:
    trainer.train()
print(f"trained in {t['seconds'] / 60:.1f} min")
````

````python
import matplotlib.pyplot as plt
trainer.save_model(OUT / f"checkpoint-{trainer.state.global_step}")
hist = [h for h in trainer.state.log_history if "loss" in h]
sp.save_json({"run": RUN, "condition": CONDITION, "seed": SEED, "model": MODEL, "hp": HP, "lora": LORA,
              "dense_until": DENSE_UNTIL, "dense_every": DENSE_EVERY if DENSE_UNTIL else None,
              "base_dtype": str(next(model.parameters()).dtype), "n_examples": len(train), "train_minutes": t["seconds"] / 60, "log_history": trainer.state.log_history,
              "provenance": sp.provenance(SEED)}, OUT / "train_log.json")
plt.plot([h["step"] for h in hist], [h["loss"] for h in hist])
plt.xlabel("step"); plt.ylabel("train loss"); plt.title(RUN)
sorted(p.name for p in OUT.glob("checkpoint-*"))
````

````python
def hub_backup(local, path_in_repo, ignore=None):
    repo = os.environ.get("SP_HUB_REPO")
    if not repo or not os.environ.get("HF_TOKEN"):
        return False
    try:
        from huggingface_hub import HfApi
        api = HfApi(token=os.environ["HF_TOKEN"])
        prefix = os.environ.get("SP_HUB_PREFIX", "")
        target = f"{prefix}/{path_in_repo}".strip("/")
        if Path(local).is_dir():
            api.upload_folder(repo_id=repo, folder_path=str(local), path_in_repo=target, ignore_patterns=ignore,
                              commit_message=f"backup {target}")
        else:
            api.upload_file(repo_id=repo, path_or_fileobj=str(local), path_in_repo=target,
                            commit_message=f"backup {target}")
        print(f"backed up {target} -> {repo}", flush=True)
        return True
    except Exception as e:
        print(f"!! hub backup failed, run continues: {type(e).__name__}: {str(e)[:200]}", flush=True)
        return False

from pathlib import Path
print("hub backup:", "on" if os.environ.get("SP_HUB_REPO") and os.environ.get("HF_TOKEN") else "off")
````

````python
# Adapters only (no optimizer state), so the private backup stays small.
hub_backup(OUT, f"runs/{RUN}", ignore=["**/optimizer.pt", "**/scheduler.pt", "**/rng_state*.pth"])
````
